```python
## Editing the series using indexing
import pandas as pd;
a = pd.Series([1,2,3,4,5,6]);
print(a);
```

```python
a[0] = 10;
a[1] = 20;
print(a); ## Here the values in the series got modifies. Here we are modifying using 
## the indexing of the series similarly we can do using the slicing
```

```python
## Editing the series using Slicing
a[1:5] = 100;
print(a);
```

```python
##Editing the series using Fancy Indexing
a[[1,2,4,5]] = 20;
print(a);
a[[1,2,4,5]] = [100,100,100,6];
print(a);
```

```python
## Now dealing with the methods common in python and Pandas
print(len(bollywood)); ## Total of 1500 movies
print(type(bollywood));
print(dir(bollywood));
print(sorted(bollywood));
```

```python
## Type conversion we generally convert the pandas series into list tuples and many other structures as per out use case
bl = list(bollywood);
print(bl);
bt = tuple(bollywood);
print(bt);
bd = dict(bollywood);
print(bd);
```

```python
## using membership operators in the Series
print('Vicky Ahuja' in bollywood); ## It returns us false because it only checks the keys not the values that is the reason
```

```python
## We can as usual use loops 
for i in bollywood.keys():
    print(bollywood[i]);
```

```python
## Arithematic Operator 
a = pd.Series([10,10,20,20,2]);
print(100 - a);
```

```python
## Relational Operators
a = pd.Series([5,10,103,332,343]);
print(a > 10);
```
