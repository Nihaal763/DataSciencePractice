```python
import numpy as np;
import pandas as pd;
subs = pd.read_csv('data/subs.csv').squeeze();
print(subs.head());
kohliruns = pd.read_csv('data/kohli_ipl.csv',index_col='match_no').squeeze();
print(kohliruns.head())
print(kohliruns.head(2));
print(kohliruns.tail());
print(kohliruns.tail(5));
```

```python
bollywood = pd.read_csv('data/bollywood.csv',index_col='movie').squeeze();
print(bollywood.sample());
print(bollywood.sample());## Generates one random row 
print(bollywood.sample(5));
```

```python
# value_counts function gives the frequency of each item in decreasing order
print(bollywood.value_counts());

```

```python
print(kohliruns.sort_values());
print(kohliruns.sort_values(ascending=False));
```

```python
## Finding highest runs made by virat kohli
kohliruns = pd.read_csv('data/kohli_ipl.csv', index_col='match_no').squeeze();
print(kohliruns.sort_values(ascending=False).head(1).values[0]);
## This does not modify the original array
```

```python
## for actual modification of the array we use methods like  sort_values(inplace = True)
kohliruns = pd.read_csv('data/kohli_ipl.csv', index_col='match_no').squeeze();
kohliruns.sort_values(inplace=True);
print(kohliruns);
```

```python
print(bollywood.sort_index());
bollywood.sort_index(inplace=True);
print(bollywood);
```
