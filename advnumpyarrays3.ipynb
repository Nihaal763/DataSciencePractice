Here we are going to see why numpy arrays are faster when compared to python list and why in terms of speed and memory


```python

a = [i for i in range(10000000)];
b = [i for i in range(10000000,20000000)];
c = [];
import time;
start = time.time();
for i in range(len(a)):
    c.append(a[i] + b[i]);

print(time.time() - start);





```

```python
import numpy as np;
a = np.arange(10000000);
b = np.arange(10000000,20000000);
start = time.time();
c = a + b;
print(time.time() - start);
```

```python
2.04/0.14
 ## So with this we can confirm that numpy arrays are way faster than the lists because of their fixed size
 ## in memory where as list are dynamic arrays and the references are stored not the arrays as continous blocks so to access and do computations on them it takes time
```

```python
## seeing in terms of memory both the list and numpy arrays how much space does they take on the RAM
a = [i for i in range(10000000)];
import sys 
print(sys.getsizeof(a));
b = np.arange(10000000,dtype = np.int32);
c = np.arange(10000000,dtype = np.int16);
print(sys.getsizeof(b));
print(sys.getsizeof(c));
```

```python
## In the previous class we had seen indexing and slicing now we will see adv Indexing
## Fancy Indexing it is nothing but we cant always find the patterns for rows and columns to get the desired elements so we use this
a = np.arange(24).reshape(6,4);
print(a);
```

```python
##Now i need to print 0,1,3,4,5 rows to get printed for me now there is no pattern for doing this
print(a[[0,1,3,4,5]]);
## Now i need only only first second and last cols not all
print(a[[0,1,3,4,5],[0,1,3]])
```

```python
## Another type of indexing Boolean indexing here we get those elements/items from the array that satisfy the given condition
a = np.random.randint(1,100,24).reshape(4,6);
print(a);
print(a > 20);
```

```python
## We need to get those elements from the array which are greater than 20
print(a[a > 20]);
## here we use the boolean masking and get only those particular elements that satisfy the condition and the result is in the form of 1D Array
```

```python
## Find all the numbers greater than 50 and are even
print(a[(a > 50) & (a % 2 == 0)]);
```

```python
## Find all the numbers that are not divisible by 7
print(a[a % 7 != 0]);
```
