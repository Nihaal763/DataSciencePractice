```python
## Creating Dataframes Using lists(2d)
import numpy as np;
import pandas as pd;
students_data = [[100,83,11],[110,92,16],[78,26,4]];
students_data = pd.DataFrame(students_data,columns=['iq','marks','package']);
print(students_data);
```

```python
## Creating dataframe using dictionary
students_data = {
    'iq' : [100,110,78],
    'marks' : [83,92,4],
    'package' : [11,16,4]
}
students_data = pd.DataFrame(students_data);
print(students_data);
```

```python
print(students_data.rename(columns={'marks': 'percentile','package': 'LPA'}));
print(students_data);## This doesn't actually change the dataframe
```

```python
## To actually change the dataframe
students_data.rename(columns={'marks': 'percentile','package': 'LPA'},inplace=True);
print(students_data);
```

```python
## Mathematical methods related to dataframes
print(students_data.sum()); ## Here it is adding in col wise to get the row wise addition use axis parameter and set it to 1
print(students_data.sum(axis = 1));
```

```python
print(movies.sum())## Here there are many cols and different data types not feasible to add
```

```python
print(students_data.mean());## Here we get the col wise mean
print(students_data.mean(axis = 1));## To get the row wise mean we do this
```

```python
## Rest of the mathematical functions are also as usual setting the axis parameter
print(students_data.median());
print(students_data.mode());
print(students_data.std());
print(students_data.var());
```
