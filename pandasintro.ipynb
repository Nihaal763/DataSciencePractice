```python
import numpy as np;
import pandas as pd;
## Series and dataframes are 2 major objects we deal here in pandas
## Now we will be dealing with Series objects
## Series are nothing but of 1d array holding data of any type
## Creating series using Strings

##Series from lists
country = ["India","Pakistan","USA"];
print(pd.Series(country));## We are creating an object of class Series type
## Here in the result it is giving the dtype as object because pandas treat Strings as objects
runs = [14,56,43,65,102,32,233];
indBatsmenruns = pd.Series(runs);
print(pd.Series(runs));

```

```python
## Creating custom indexes of our own type
score = [102,31,88,93];
batsmen = ['Warner','Khawaja','Labuschagne','Smith'];
print(pd.Series(score,index=batsmen));
```

```python
## Also we could give the name to the object as well
print(pd.Series(score,index=batsmen,name = 'Aussiebatsscore'));
ausBatsscore = pd.Series(score,index=batsmen,name = 'Aussiebatsscore');
print(ausBatsscore);
## Both of them actually print the same thing but any changes will reflect to the variable pointing it

```

```python
## Creating series using dictionaries
batsmenscoreAus = {
    'Warner' : 102,
    'Khawaja' : 31,
    'Labuschagne' : 88,
    'Smith' : 93
}
print(pd.Series(batsmenscoreAus));## While creating this custom indexes will be replaced by key in the dictionary
```

```python
## Attributes of Series
print(ausBatsscore.size);
print(ausBatsscore.dtype);
print(ausBatsscore.name);
print(ausBatsscore.is_unique);
print(ausBatsscore.index);# Returns us the index object
runs = [14,56,43,65,102,32,233];
indBatsmenruns = pd.Series(runs);
print(indBatsmenruns.index);
print(ausBatsscore.values);## Returns all the values in the form of numpy array
print(type(ausBatsscore.values));
```
