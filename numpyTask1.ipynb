Q-3Write a function to create a 2d array with 1 on the border and 0 inside. Take 2-D array shape as (a,b) as parameter to function.

```python
a = int(input("Enter the value of a :"));
b = int(input("Enter the value of b :"));
c = np.zeros((a,b));
def ansfn(a,b):
    c[0,:] = 1;
    c[a-1,:] = 1;
    c[:,0] = 1;
    c[:,b-1] = 1;
ansfn(a,b);
print(c);
```

Q-7: Consider a random integer (in range 1 to 100) vector with shape (10,2) representing coordinates, and coordinates of a point as array is given. Create an array of distance of each point in the random vectros from the given point. Distance array should be interger type.

```python
a = np.random.randint(1,100,20).reshape(10,2);
point = np.array([2,3]);
print(a);
print(point);
def distance(coordinates,point):
    darray = np.array([]);
    darray = np.append(darray,np.sqrt(np.sum((coordinates-point)**2 , axis = 1)));
    return darray.astype(np.int32);
print(distance(a,point));
```
