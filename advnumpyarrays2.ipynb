```python
import numpy as np;
a1 = np.arange(10);
a2 = np.arange(12).reshape(3,4);
a3 = np.arange(8).reshape(2,2,2);
print(a1 , a2 , a3);
```

```python
print(a2);
```

```python
print(a2[1,3]);
print(a2[1][2]);
## In either of the ways we can access the elements it does not matter which one we are using
print(a2[-1,-1]);
print(a2[-1,0]);
```

```python
print(a3);
```

```python
print(a3[0,0,0]);
print(a3[0,0,1]);
print(a3[1,1,0]);
```

```python
## Previously with using indexing we were able to only get 1 particular element
## But now with using slicing we can get multiple elements simultaneously
print(a1);
```

```python
print(a1[0:5]);
print(a1[0:5:2]);
```

```python
print(a2);
```

```python
##Printing 0th column
print(a2[:,0]);
## Printing 0th row
print(a2[0,:]);
## Printing all the corner elements
print(a2[::2,::3]);
print(a2[::2,1::2]);
print(a2[1::2,::3]);
print(a2[:2,1::]);
```

```python
print(a3);
```

```python
a3 = np.arange(27).reshape(3,3,3);
print(a3);
```

```python
print(a3[1,:,1]);
print(a3[2,1:,1:]);
print(a3[::2,0,::2]);
```

```python
## Finding the transpose of the matrix
print(np.transpose(a2));
```

```python
## There is a function called ravel invoked on the numpy array it converts the numpy array of n dim into the 1d array
print(a2.ravel());
print(a3.ravel());
```

```python
## Horizontal Stacking combining two matrices of same shape(size) horizontally so only cols increase no of rows remains the same
a = np.arange(8).reshape(2,4);
b = np.random.random((2,4));
b = b.astype(np.int32);
c = np.hstack((a,b));
print(c);
print(np.hsplit(c , 4));## After stacking if we wanted to split at any point of time we can do that so it in any equal no of parts
```

```python
## Vertical Stacking is completely the opposite of horizontal stacking
a = np.arange(8).reshape(2,4);
b = np.random.random((2,4));
b = b.astype(np.int32);
c = np.vstack((a,b));
print(c);
print(np.vsplit(c , 2));
```
