```python
import numpy as np;
import pandas as pd;
import sys;

a = pd.Series([10,120,230,40]); ## Changing the datatype of the Series because of small data type and conserving the space
print(a);
print(sys.getsizeof(a));
a = a.astype('int8');
print(sys.getsizeof(a));
print(a.dtype);
```

```python
## between method returns us a boolean series if the item in the series satisfies the given condition
kohliruns = pd.read_csv('data/kohli_ipl.csv',index_col='match_no').squeeze();
print(kohliruns.head(2));
print(kohliruns[kohliruns.between(51,100)]);
```

```python
## Clip Method is used for making our data in the given range for easy data analysis
subs = pd.read_csv('data/subs.csv').squeeze();
print(subs);
print(subs.clip(0,200));## Here any number greater than 200 is given the value 200 and less than that is kept same
```

```python
## for removing duplicates we invoke a method on the series object called drop_duplicates
a = pd.Series([1,2,2,22,4,3,1,1,2,3,3]);
print(a.drop_duplicates());## It does not modify the original Series
print(a.drop_duplicates(keep='last'));## Removes the duplicates that are occuring first
print(a);
print(a.duplicated().sum());## Gives the no of non duplicates in the series
```

```python
b = pd.Series([10,20,np.nan,np.nan,3,np.nan,2,11]);
print(b.size);## directly returns off the size 
print(b.count());## Count returns only valid datatypes values in the Series
```

```python
print(b.isnull().sum());## Returns the no of null or nan values in the series
```

```python
print(b.dropna());
```

```python
print(b.fillna(10));## ALl the null or isNan values are replaced by the value we put here
```

```python
## to check whether the values or the indexes exist in the series we use the method isin()
print(kohliruns[kohliruns.isin([49,99,90])]);
```

```python
bollywood = pd.read_csv('data/bollywood.csv').squeeze();
print(bollywood.head(3));
bollywood = bollywood.apply(lambda x:x.str.split());
print(bollywood);
```

```python
## The main use of apply method is
a = pd.Series([1,2,3,4,5,23,33,5,6,7]);
print(a.apply(lambda x:x*2));## does not actually modify the array
print(a);
```
