```python
## Scalar operations
import numpy as np;
a1 = np.arange(12).reshape(3,4);
a2 = np.arange(12,24).reshape(3,4);

print(a1 + 2);
print(a1 * 2);
print(a1 / 2);
print(a1 ** 2);
```

```python
## Relational operations
import numpy as np;
a1 = np.arange(12).reshape(3,4);
a2 = np.arange(12,24).reshape(3,4);

print(a2 > 100);
print(a2 < 20);
```

```python
import numpy as np;
a1 = np.arange(12).reshape(3,4);
a2 = np.arange(12,24).reshape(3,4);

## Vector operations if dimensions of the array are same we can directly add
print(a1 + a2);
print(a1 / a2);
```

```python
## Functions to do the mathematical operations we dont need to do anything

import numpy as np;
a1 = np.arange(12).reshape(3,4);
a2 = np.arange(12,24).reshape(3,4);
## for finding the max number in the numpy array we use np.max function
print(np.max(a2));
## Similarly there is the min number in the numpy array we use np.min function
print(np.min(a2));
## Here we can even find the max and min in each row and col
## for row -> 1 , col -> 0
print(np.max(a2 , axis = 1));
```

```python
## Statistical related operations
import numpy as np;
a1 = np.arange(12).reshape(3,4);
a2 = np.arange(12,24).reshape(3,4);
print(np.mean(a2));
print(np.median(a2));
print(np.std(a2));
print(np.var(a2));
```

```python
## Trigonometric functions

import numpy as np;
a1 = np.arange(12).reshape(3,4);
a2 = np.arange(12,24).reshape(3,4);

np.sin(a1);
np.cos(a2);
```

```python
## Dot product(We find the dot product between the 2 matrices(a,b & b,c) then dimension of matrix is a,c)
import numpy as np;
a1 = np.arange(12).reshape(3,4);
a2 = np.arange(12,24).reshape(3,4);
a3 = np.arange(24,36).reshape(4,3);

print(np.dot(a3,a2));
```

```python
## Log and exponents
import numpy as np;
a1 = np.arange(12).reshape(3,4);
a2 = np.arange(12,24).reshape(3,4);

print(np.log(a2));
print(np.exp(a2));
```

```python
## Round , floor and ceil functions
print(np.round(np.random.random((3,4))*100));
print(np.floor(np.random.random((3,4))*100));
print(np.ceil(np.random.random((3,4))*100));
```
