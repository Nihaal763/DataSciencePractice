```python
import numpy as np;
import pandas as pd;
ipl = pd.read_csv('data/ipl-matches.csv');
movies = pd.read_csv('data/movies.csv');

```

```python
print(movies);
```

```python
## Fetching columns from a dataframe
print(movies['title_x']);## Printing a particular column(indirectly we are printiing a series)
print(type(movies['title_x']));
```

```python
## Fetching multiple columns from a dataframe
print(movies[['title_x','imdb_id','title_y']]);## If we are fetching multiple columns then it will be dataframe if we are fetching only a single col then it is a series
print(type(movies[['title_x','imdb_id','title_y']]));
```

```python
## Fetching rows from a dataframe
studentsDict = {
    'name' : ['nihaal' ,'Karthikeya','Akhil','Sameer'],
    'iq' : [100,125,92,84],
    'percentile' : [85,96,78,86],
    'LPA' : [65,83,72,70]
}
students_data = pd.DataFrame(studentsDict);
print(students_data);
students_data.set_index('name',inplace=True);
print(students_data);
```

```python
## iloc ->> Searches using index positions
## loc -->> Searches using index labels
print(movies.iloc[0]);## we are generating only the 1st row(indirectly a series
print(type(movies.iloc[0]));
```

```python
print(type(movies.loc[[1,2,3,4,0]]));## Since we are fetching multiple rows it becomes a dataframe
```

```python
## there is also other method of selecting multiple in iloc using slicing method
print(movies.iloc[1:6]);
```

```python
## We could also use fancy indexing in iloc
print(movies.iloc[[0,1,4,2,3,12]]);
```

```python
## Now seeing about loc
print(students_data.loc['nihaal']);
```

```python
print(students_data.loc['nihaal':'Sameer']);
print(students_data.loc['nihaal':'Sameer':2]);
print(type(students_data.loc['nihaal':'Sameer']));## As we are selecting multiple rows so the output will the combination of multiple series i.e a dataframe
```

```python
## Even fancy indexing also works here also
print(students_data.loc[['nihaal','Karthikeya']]);
```

```python
print(students_data.iloc[0]);## this is working because although we had set an index pandas has its own index for it
```

```python
## Selecting both rows and columns
print(movies.iloc[0:3,0:3]);
```

```python
## We can do the above using loc also
print(movies.loc[0:3,['title_x','poster_path']]);
```

```python
## One point to remember is that even if we set our index in dataframes pandas has its own indexes running at the back
```

```python
## Filtering a dataframe
print(ipl.sample(1));
```

```python
## Find all the ipl final winners
print();
finals = ipl['MatchNumber'] == 'Final';
finalWinners = ipl[finals];
print(finalWinners[['Season','WinningTeam']]);
```

```python
## for getting the output in one line
print(ipl[ipl['MatchNumber'] == 'Final'][['Season','WinningTeam']]);
```

```python
## How many superover finishes have occured
print((ipl['SuperOver']=='Y').sum());
print(type(ipl['SuperOver']=='Y'));
## Other way is using the shape
print(ipl[ipl['SuperOver']=='Y'].shape);
print(ipl[ipl['SuperOver']=='Y'].shape[0]);
```

```python
## How many matches has csk won in Kolkata
print(ipl[(ipl['City'] == 'Kolkata') & (ipl['WinningTeam'] == 'Chennai Super Kings')].shape[0]);
```

```python
## toss winner is match winner in percentage
print((ipl[ipl['TossWinner'] == ipl['WinningTeam']].shape[0]) * 100/(ipl.shape[0]));
```

```python
### Now filtering in movies 
## Movies with rating higher than 8 and votes > 10000
print(movies[(movies['imdb_rating'] > 8) & (movies['imdb_votes'] > 10000)]['title_x']);
print(movies[(movies['imdb_rating'] > 8) & (movies['imdb_votes'] > 10000)].shape[0]);
```

```python
## Action movies with rating higher than 7.5
print(movies[(movies['imdb_rating'] > 7.5) & (movies['genres'].str.contains('Action'))].shape[0]);
print(movies[(movies['imdb_rating'] > 7.5) & (movies['genres'].str.split('|').apply(lambda x:'Action' in x))].shape[0]);
```

```python
## Adding new column in the datasets
movies['country'] = 'India'
print(movies.sample(1));
```

```python
movies['leadActor'] = movies['actors'].str.split('|').apply(lambda x:x[0]); ## So there are some missing or null values so there is float object there so using dropna
```

```python
## Changing datatype of the column indirectly serieses
import sys;
print(sys.getsizeof(ipl['ID']));
ipl['ID'] = ipl['ID'].astype('int16');
print(sys.getsizeof(ipl['ID']));
```

```python
## There is actually an object in Pandas called category it is similar to non null object but occupies less than compared to it
print(sys.getsizeof(ipl['Venue']));
ipl['Venue'] = ipl['Venue'].astype('category');
print(sys.getsizeof(ipl['Venue']));
```
